```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img1=cv2.imread("/Users/pritam/Downloads/20261016_012026.jpg")
```

```python
plt.imshow(img1)
```

```python
img2=cv2.imread("/Users/pritam/Downloads/20261016_012226.jpg")
```

```python
plt.imshow(img2)
```

```python
add_img=cv2.add(img1,img2)
plt.imshow(add_img)
```

```python
weighted_img=cv2.addWeighted(img1,0.5,img2,0.4,0)
```

```python
plt.imshow(weighted_img)
```

```python
weighted_img=cv2.addWeighted(img1,1,img2,1,0)
plt.imshow(weighted_img)
```

```python
weighted_img=cv2.addWeighted(img1,0.6,img2,0.4,0)
plt.imshow(weighted_img)
```

```python
weighted_img=cv2.addWeighted(img1,0.6,img2,0.4,10)
plt.imshow(weighted_img)
```

```python
weighted_img=cv2.addWeighted(img1,0.6,img2,0.4,100)
plt.imshow(weighted_img)
```

```python
sub_img=cv2.subtract(img1,img2)
```

```python
plt.imshow(sub_img)
```

```python
sub_img2=cv2.cvtColor(sub_img,cv2.COLOR_BGR2RGB)  #BGR to RGB converstion
plt.imshow(sub_img2)
```

```python
anding=cv2.bitwise_and(img2,img1)
plt.imshow(anding)
```

```python
ORing=cv2.bitwise_or(img2,img1)
plt.imshow(ORing)
```

```python
xoring=cv2.bitwise_xor(img2,img1)
plt.imshow(xoring)
```

```python
noting=cv2.bitwise_not(img2,img1)
plt.imshow(noting)
```
