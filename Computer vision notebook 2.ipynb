```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("/Users/pritam/Downloads/_DSC4117.JPG") #reads as BGR
```

```python
img.shape
```

```python
type(img)
```

```python
plt.imshow(img)
```

```python
pic=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)  #BGR to RGB converstion
plt.imshow(pic)
```

```python
pic_resize=cv2.resize(pic,(100,100))
```

```python
plt.imshow(pic_resize)
```

```python
pic2_resize=cv2.resize(pic,(0,0),pic,0.5,0.5)
```

```python
plt.imshow(pic2_resize)
```

```python
pic_flip=cv2.flip(pic2_resize,0)
plt.imshow(pic_flip)
```

```python
pic_flip=cv2.flip(pic2_resize,1)
plt.imshow(pic_flip)
```

```python
pic_flip=cv2.flip(pic2_resize,-1)
plt.imshow(pic_flip)
```

```python
cv2.imwrite('saved.png',pic2_resize)
```
