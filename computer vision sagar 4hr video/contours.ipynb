```python
import cv2
```

```python
image=cv2.imread("triangle.jpeg")
```

```python
grey=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

```python
_,thresh=cv2.threshold(grey,240,255,cv2.THRESH_BINARY_INV)
```

```python
contours,heirarchy=cv2.findContours(thresh,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)
```

```python
cv2.drawContours(image,contours,-1,(0,255,0),3)
```

```python
for contour in contours:
    # Ignore very small contours caused by noise
    if cv2.contourArea(contour) < 100:
        continue

    approx = cv2.approxPolyDP(contour, 0.08 * cv2.arcLength(contour, True), True)
    corners = len(approx)

    # Initialize default shape_name to prevent NameError later
    shape_name = "Unknown"

    if corners == 3:
        shape_name = "Triangle"
    elif corners == 4:
        shape_name = "Rectangle" # or Square depending on aspect ratio
    elif corners == 5:
        shape_name = "Pentagon"
    elif corners == 6:
        shape_name = "Hexagon"
    elif corners == 7:
        shape_name = "Heptagon"
    elif corners == 8:
        shape_name = "Octagon"
    elif corners > 8: # Circles typically have many approximated corners
        shape_name = "Circle"
    else:
        continue

    cv2.drawContours(image, [approx], 0, (0, 255, 0), 2)
    x = approx.ravel()[0]
    y = approx.ravel()[1] - 10  # Fixed TypeError: changed -[10] to - 10

    cv2.putText(image, shape_name, (x, y), cv2.FONT_HERSHEY_SIMPLEX,0.6, (0, 255, 0), 2)
```

```python
cv2.imshow("Contours",image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
