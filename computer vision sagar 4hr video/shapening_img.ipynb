```python
import cv2
import numpy as np
```

```python
image=cv2.imread("/Users/pritam/anaconda_projects/computer vision sagar 4hr video/shapening practice.webp")
```

```python
sharpening_kernal=np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])
```

```python
sharpend_image=cv2.filter2D(image,-1,sharpening_kernal)
```

```python
cv2.imshow("original img",image)
cv2.imshow("shapening image",sharpend_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
